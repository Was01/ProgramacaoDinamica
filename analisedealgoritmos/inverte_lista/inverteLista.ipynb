```python
def invertelista1(lista):
  tamanho=len(lista)
  limite=tamanho//2
  for i in range (limite):
    lista[i],lista[tamanho-1-i]=lista[tamanho-1-i],lista[i]
  return lista
```

```python
minha_lista=[1,5,7,8,9]
invertelista1(minha_lista)
```

```python
def invertelista2(lista):
  nova_lista=[]
  tamanho=len(lista)
  for i in range(tamanho):
    nova_lista.append(lista[tamanho-1-i])
  return nova_lista
```

```python
nova_lista=[23,6,7,-3]
invertelista2(nova_lista)
```
